```python
import pandas as pd
#cargo la base de datos
df_pi = pd.read_csv('base_datos_restaurantes_USA_v2.csv', encoding = 'utf-8')
#muestro la informacion de la base de datos
df_pi.info()
```

```python
#Visualizo las filas de la base de datos
df_pi
```

```python
#Analizo la cantidad de filas y columnas que están vacias.
df_pi.isnull().mean() * 100
```

```python
#Utilizo el metodo describe para obtener un resumen estadistico
df_pi.describe()
```

```python
#comienzo el filtrado por edades.
df_pi = df_pi[(df_pi['edad'] >= 18) & (df_pi['edad'] <= 80)]
```

```python
#Utilizo el metodo describe para verificar el filtrado por edades.
df_pi.describe()
```

```python
#Ahora voy a completar las filas vacias de edad con la media de edad. Luego verifico.

df_pi['edad'] = df_pi['edad'].fillna(df_pi['edad'].mean())

df_pi.isnull().mean() * 100
```

```python
"""El mínimo de gasto es 0, Lo cual es extraño, ya que se supone que las personas gastan algo en comida. Por lo tanto, voy eliminar los valores menores a 1
"""

df_pi = df_pi[df_pi['promedio_gasto_comida'] >= 1]

#verifico el nuevo valor minimo de gasto en comida
df_pi['promedio_gasto_comida'].min()
```

```python
#completo los valores vacios de promedio_gasto_comida con la media de gasto en comida. Luego verifico.

df_pi['promedio_gasto_comida'] = df_pi['promedio_gasto_comida'].fillna(df_pi['promedio_gasto_comida'].mean())

df_pi.isnull().mean() * 100
```

```python
"""Comienzo el analisis de las preferencias alimenticias. Primero, verifico los valores unicos."""

print(df_pi['preferencias_alimenticias'].unique())
```

```python
"""Tomé la decisión de crear una nueva categoria llamada "No especificado" para aquellos valores vacios en preferencias alimenticias, ya que no quiero perder información de esas filas, pero tampoco quiero asignarles una categoría que no les corresponda."""

df_pi ['preferencias_alimenticias'] = df_pi ['preferencias_alimenticias'].fillna('No especificado')
```

```python
print(df_pi['preferencias_alimenticias'].unique())
```

```python
"""Telefono_contacto y correo_electronico tienen un 50% de valores vacios y no aportan información relevante
para el análisis. Por lo tanto, voy a eliminar esas columnas."""

df_pi = df_pi.drop(columns = ['telefono_contacto', 'correo_electronico'])
```

```python
df_pi.isnull().mean() #Verifico que si hay valores nulos.
```

Distribución de personas según estrato socioeconómico.

```python
#Distribución de personas según estratos socioeconómicos

resumen_estratos = df_pi ['estrato_socioeconomico'].value_counts().reset_index()

porcentaje_estratos = (df_pi['estrato_socioeconomico'].value_counts(normalize=True).values * 100).round(2)

resumen_estratos.columns = ['Estrato' , 'Cantidad']

print (resumen_estratos)
print (porcentaje_estratos)
```

```python
plt.figure(figsize= (15 , 8))
plt.bar(resumen_estratos ['Estrato'] , 
        resumen_estratos ['Cantidad'],
        color ='skyblue')

plt.title ('Distribución de personas por estrato socioeconómico')
plt.xlabel ('Estrato')
plt.ylabel ('C. Personas')

for i, v in enumerate(resumen_estratos['Cantidad']):
    # v ahora es solo el número (int), así que la suma y el str(v) funcionarán bien
    plt.text(i, v+ 100, str(v), ha='center', fontweight='bold')
"""
i: índice, pide que se coloque encima de la barra
v: es el valor obtenido, lo que quiero escribir sobre la barra
v + 100: hace que esté encima de la barra pero un poco mas arriba así no queda pegado al borde superior de ella.
srt(v): el módulo matplotlib solicita string. como V es un número lo convierto.
ha: solicito que se coloque el texto en el centro de la barra.
fontweigh: solicito el texto en negrita.
"""
plt.show ()
```

```python
#Estratos por ciudad.

plt.figure(figsize=(15, 8))
sns.countplot(x='ciudad_residencia',hue='estrato_socioeconomico',  data=df_pi)
"""
con countplot yo no defino el eje Y. y hue divide cada ciudad en sub barras de diferentes colores.Sin hue 
tendría un gráfico por ciudad.
"""
plt.title('Distribución de Estratos por Ciudad')
plt.ylabel ('C. Personas')
plt.xticks(rotation=45)
plt.show()
```
